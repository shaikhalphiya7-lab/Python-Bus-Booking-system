```python
# Project: Bus Reservation System
# Name: Alphiya
# As per PDF - SwiftTravel

class Passenger:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Booking:
    def __init__(self, booking_id, passenger, bus_id, route, seat, fare):
        self.booking_id = booking_id
        self.passenger = passenger
        self.bus_id = bus_id
        self.route = route
        self.seat = seat
        self.fare = fare
        self.status = "CONFIRMED"

# data as per PDF page 1
buses = {
    "B101": {"route": "Pune -> Mumbai", "time": "08:00 AM", "fare": 450, "seats": 10, "booked": {}},
    "B102": {"route": "Pune -> Nashik", "time": "09:30 AM", "fare": 350, "seats": 12, "booked": {}}
}
bookings = {}
next_id = 1001

def calculate_fare(base_fare, age):
    # as per PDF 6.4
    if age < 12:
        return base_fare * 0.5
    elif age >= 60:
        return base_fare * 0.7
    else:
        return base_fare

def view_buses():
    print("\n---------------------------------------------------------")
    print("ID ROUTE TIME FARE SEATS")
    print("---------------------------------------------------------")
    for bid in buses:
        b = buses[bid]
        avl = b["seats"] - len(b["booked"])
        print(f"{bid} {b['route']:<12} {b['time']} Rs.{b['fare']} {avl}")
    print("---------------------------------------------------------")

def view_seats():
    # PDF 6.2 exact format
    bid = input("Enter Bus ID: ").strip().upper()
    if bid not in buses:
        print("Invalid Bus ID")
        return
    b = buses[bid]
    print(f"\n BUS {bid}")
    print()
    for i in range(1, b["seats"]+1):
        if i in b["booked"]:
            print(f" [X] ", end="")
        else:
            print(f" [{i:02d}]", end="")
        if i % 2 == 0:
            print()
    # show available and booked list
    available = []
    booked = []
    for i in range(1, b["seats"]+1):
        if i in b["booked"]:
            booked.append(f"{i:02d}")
        else:
            available.append(f"{i:02d}")
    print(f"\nAvailable: {' '.join(available)}")
    print(f"Booked: {' '.join(booked)}")

def book_ticket():
    global next_id
    try:
        name = input("Enter passenger name: ").strip().title()
        age = int(input("Enter age: ").strip())
        journey = input("Enter journey ID: ").strip().upper()
        seat = int(input("Select seat: ").strip())

        if journey not in buses:
            print("Invalid Journey ID")
            return
        b = buses[journey]
        if seat < 1 or seat > b["seats"]:
            print("Invalid seat number")
            return
        if seat in b["booked"]:
            print("Seat already booked, choose another")
            return

        fare = calculate_fare(b["fare"], age)
        p = Passenger(name, age)
        booking_id = f"ST{next_id}"
        next_id += 1

        booking = Booking(booking_id, p, journey, b["route"], seat, fare)
        bookings[booking_id] = booking
        b["booked"][seat] = name

        # file handling - PDF requirement 9
        with open("bookings_Alphiya.txt", "a") as f:
            f.write(f"{booking_id},{name},{age},{journey},{seat},{fare}\n")

        print("\n========================================")
        print(" BOOKING CONFIRMED")
        print("========================================")
        print(f"Booking ID : {booking_id}")
        print(f"Passenger : {name}")
        print(f"Journey : {journey}")
        print(f"Route : {b['route']}")
        print(f"Seat : {seat:02d}")
        print(f"Fare : Rs.{fare}")
        print(f"Status : CONFIRMED")
        print("========================================")

    except ValueError:
        print("Invalid input. Age must be a number. Please try again.")

def cancel_ticket():
    bid = input("Enter Booking ID: ").strip().upper()
    if bid in bookings:
        data = bookings[bid]
        print(f"\nBooking found.")
        print(f"Passenger: {data.passenger.name}")
        print(f"Seat: {data.seat:02d}")
        confirm = input("Cancel booking? (yes/no): ").strip().lower()
        if confirm == "yes":
            buses[data.bus_id]["booked"].pop(data.seat)
            del bookings[bid]
            print("Booking cancelled successfully.")
            print(f"Seat {data.seat:02d} is now available.")
    else:
        print("Booking ID not found")

def view_booking():
    bid = input("Enter Booking ID: ").strip().upper()
    if bid in bookings:
        d = bookings[bid]
        print("\n========================================")
        print(" BOOKING DETAILS")
        print("========================================")
        print(f"Booking ID : {d.booking_id}")
        print(f"Passenger : {d.passenger.name}")
        print(f"Journey : {d.bus_id}")
        print(f"Route : {d.route}")
        print(f"Seat : {d.seat:02d}")
        print(f"Fare : Rs.{d.fare}")
        print(f"Status : {d.status}")
        print("========================================")
    else:
        print("Booking not found")

# main menu - PDF page 2
print("=====================================")
print(" BUS RESERVATION SYSTEM")
print("=====================================")
while True:
    print("\n1. View Buses")
    print("2. View Available Seats")
    print("3. Book Ticket")
    print("4. Cancel Ticket")
    print("5. View Booking")
    print("6. Exit")
    ch = input("Enter your choice: ").strip()
    if ch == "1":
        view_buses()
    elif ch == "2":
        view_seats()
    elif ch == "3":
        book_ticket()
    elif ch == "4":
        cancel_ticket()
    elif ch == "5":
        view_booking()
    elif ch == "6":
        print("Thank you Alphiya! Visit again.")
        break
    else:
        print("Invalid choice")
```
